# creating ccoonditional column from more than 2 choices:np.select()

```python
conditions=[
    df_laptop['Final Price']>3000,
           (df_laptop['Final Price']>2000)&(df_laptop['Final Price']<=3000),
           (df_laptop['Final Price']>800)& (df_laptop['Final Price']<=2000),
          (df_laptop['Final Price']<=800)
                                            ]
values=['Too Expensive','Expensive','Affordable','Cheap']
```

```python
len(conditions), len(values)
```

```python
for condition in conditions:
       print(condition.dtype)
```

```python
df_laptop['price_tier']=np.select(conditions,values,default='Uncategorized')
```
